# Level 1 - Task 2: Descriptive Analysis

## Objective
The objective of this task is to perform descriptive statistical analysis on the mobile price range dataset. We calculate the mean, median, mode, and standard deviation of numerical features and analyze RAM, Battery Power, and Internal Memory distributions.

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("MobilePriceRange.csv")
df.head()
```

```python
df.describe()
```

```python
print("Mean")
print(df.mean(numeric_only=True))
```

```python
print("Median")
print(df.median(numeric_only=True))
```

```python
print("Mode")
print(df.mode().iloc[0])
```

```python
print("Standard Deviation")
print(df.std(numeric_only=True))
```

```python
print("RAM Statistics")
print(df["RAM"].describe())
```

```python
print("Battery Power Statistics")
print(df["BatteryPower"].describe())
```

```python
print("Internal Memory Statistics")
print(df["InternalMemory"].describe())
```

```python
features = ["RAM", "BatteryPower", "InternalMemory"]

summary = pd.DataFrame({
    "Mean": df[features].mean(),
    "Median": df[features].median(),
    "Mode": df[features].mode().iloc[0],
    "Std Dev": df[features].std()
})

summary
```

## Observations

- RAM values show a wide range, indicating significant variation in device performance.
- Battery Power is distributed across low and high capacities, representing different categories of smartphones.
- Internal Memory varies considerably, showing multiple storage configurations.
- The standard deviation indicates noticeable variability in these numerical features.
- These features are expected to play an important role in predicting the mobile price range.
