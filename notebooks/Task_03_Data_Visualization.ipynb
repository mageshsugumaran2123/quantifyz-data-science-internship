# 📱 Mobile Price Range Prediction

## Level 1 - Task 3: Data Visualization

### Objective
The objective of this task is to visualize the distribution and relationships of important features in the Mobile Price Range dataset using histograms, boxplots, count plots, correlation heatmaps, and pair plots.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

```python
df = pd.read_csv("MobilePriceRange.csv")

df.head()
```

```python
print("Dataset Shape:", df.shape)
print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])
```

```python
import os
import matplotlib.pyplot as plt

# Create the folder if it doesn't exist
os.makedirs("../images/histograms", exist_ok=True)

# Create RAM histogram
plt.figure(figsize=(8, 5))

plt.hist(df["RAM"], bins=20, edgecolor="black")

plt.title("Distribution of RAM")
plt.xlabel("RAM")
plt.ylabel("Frequency")

plt.tight_layout()

# Save the graph
plt.savefig("../images/histograms/ram_distribution.png")

# Display the graph
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["BatteryPower"], bins=20, edgecolor="black")

plt.title("Distribution of Battery Power")
plt.xlabel("Battery Power")
plt.ylabel("Frequency")

plt.tight_layout()

plt.savefig("../images/histograms/battery_power_distribution.png")
plt.show()
```

```python
import os
import matplotlib.pyplot as plt
import seaborn as sns

features = [
    "BatteryPower",
    "InternalMemory",
    "ClockSpeed",
    "NoOfCores",
    "RAM",
    "TalkTime",
    "MobileDepth",
    "MobileWeight",
    "ScreenHeight",
    "ScreenWidth",
    "FrontCameraPx",
    "RearCameraPX",
    "PxHeight",
    "PxWeight"
]

# Create folder if it doesn't exist
os.makedirs("../images/boxplots", exist_ok=True)

# Create boxplot
plt.figure(figsize=(15, 8))

sns.boxplot(data=df[features])

plt.title("Boxplot of Numerical Features")
plt.xlabel("Features")
plt.ylabel("Values")
plt.xticks(rotation=45)

plt.tight_layout()

# Save the visualization
plt.savefig("../images/boxplots/numerical_features_boxplot.png")

# Display
plt.show()
```

```python
import os
import matplotlib.pyplot as plt
import seaborn as sns

# Create the folder if it doesn't exist
os.makedirs("../images/comparison", exist_ok=True)

# Create count plot
plt.figure(figsize=(8, 5))

sns.countplot(x="PriceRange", data=df)

plt.title("Distribution of Mobile Price Range")
plt.xlabel("Price Range")
plt.ylabel("Number of Mobile Phones")

plt.tight_layout()

# Save the graph
plt.savefig("../images/comparison/price_range_countplot.png")

# Display the graph
plt.show()
```

```python
import os
import matplotlib.pyplot as plt
import seaborn as sns

# Create the heatmaps folder if it doesn't exist
os.makedirs("../images/heatmaps", exist_ok=True)

# Select only numerical columns
correlation = df.select_dtypes(include="number").corr()

# Create correlation heatmap
plt.figure(figsize=(15, 10))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Numerical Features")

plt.tight_layout()

# Save the heatmap
plt.savefig("../images/heatmaps/correlation_heatmap.png", dpi=300)

# Display
plt.show()
```

```python
import os
import seaborn as sns
import matplotlib.pyplot as plt

# Create the folder if it doesn't exist
os.makedirs("../images/comparison", exist_ok=True)

# Select features from your actual dataset
pair_features = [
    "BatteryPower",
    "InternalMemory",
    "PxHeight",
    "PxWeight",
    "RAM",
    "PriceRange"
]

# Create pair plot
pair_plot = sns.pairplot(
    df[pair_features],
    hue="PriceRange"
)

# Save the pair plot
pair_plot.savefig(
    "../images/comparison/pair_plot.png",
    dpi=300
)

plt.show()
```

```python
# Create a numerical version of PriceRange for correlation analysis
price_mapping = {
    "Low": 0,
    "Medium": 1,
    "High": 2
}

df["PriceRange_Code"] = df["PriceRange"].map(price_mapping)

# Calculate correlation
correlation = df.select_dtypes(include="number").corr()

# Get correlation with PriceRange
price_correlation = correlation["PriceRange_Code"].sort_values(
    ascending=False
)

price_correlation
```
