```python
connectivity_columns = [
    col for col in ["3G", "4G", "WiFi", "Bluetooth"]
    if col in df.columns
]

print("Connectivity columns found:")
print(connectivity_columns)
```

```python
connectivity = df.groupby("PriceRange")[
    connectivity_columns
].mean()

connectivity
```

```python
os.makedirs("../images/comparison", exist_ok=True)

connectivity.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Connectivity Features Across Price Ranges")
plt.xlabel("Price Range")
plt.ylabel("Availability Proportion")
plt.xticks(rotation=0)
plt.legend(title="Feature")

plt.tight_layout()

plt.savefig(
    "../images/comparison/connectivity_by_price_range.png",
    dpi=300
)

plt.show()
```

```python
specification_summary = df.groupby("PriceRange").agg({
    "RAM": "mean",
    "BatteryPower": "mean",
    "InternalMemory": "mean",
    "NoOfCores": "mean",
    **{col: "mean" for col in connectivity_columns}
})

specification_summary
```

```python
continuous_features = [
    "RAM",
    "BatteryPower",
    "InternalMemory",
    "NoOfCores"
]

for feature in continuous_features:
    averages = df.groupby("PriceRange")[feature].mean()

    print(f"\n{feature}")
    print("Highest average:", averages.idxmax())
    print("Highest value:", round(averages.max(), 2))
```

## Key Observations

### RAM
RAM distribution was compared across the four price ranges. The analysis helps determine whether higher-priced smartphones generally have greater RAM capacity.

### Battery Power
Battery capacity was analyzed across price categories. Differences in average battery power can indicate whether battery capacity is associated with mobile price.

### Internal Memory
Internal memory was compared across price ranges to understand the relationship between storage capacity and price category.

### CPU Cores
The number of CPU cores was analyzed across price ranges to determine whether higher-priced smartphones tend to have more processing cores.

### Connectivity
The availability of 3G, 4G, WiFi, and Bluetooth was compared across price ranges using their average availability proportions.

### Overall Observation
The specification summary provides a comparative view of hardware and connectivity characteristics across Low, Medium, High, and VeryHigh price categories.

```python
connectivity_columns = [
    col for col in ["3G", "4G", "WiFi", "Bluetooth"]
    if col in df.columns
]

print("Connectivity columns found:")
print(connectivity_columns)
```

```python
connectivity = df.groupby("PriceRange")[connectivity_columns].mean()

connectivity
```

```python
connectivity_percentage = connectivity * 100

connectivity_percentage.round(2)
```

```python
import os
import matplotlib.pyplot as plt
import seaborn as sns

os.makedirs("../images/comparison", exist_ok=True)

connectivity_percentage.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Connectivity Features Across Price Ranges")
plt.xlabel("Price Range")
plt.ylabel("Availability (%)")
plt.xticks(rotation=0)
plt.legend(title="Feature")

plt.tight_layout()

plt.savefig(
    "../images/comparison/connectivity_by_price_range.png",
    dpi=300
)

plt.show()
```

```python
specification_columns = [
    "RAM",
    "BatteryPower",
    "InternalMemory",
    "NoOfCores"
]

specification_summary = df.groupby("PriceRange")[specification_columns].mean()

specification_summary.round(2)
```

```python
final_specification_summary = pd.concat(
    [
        specification_summary,
        connectivity_percentage
    ],
    axis=1
)

final_specification_summary.round(2)
```

```python
for feature in specification_columns:
    averages = df.groupby("PriceRange")[feature].mean()

    print(f"\n{feature}")
    print("Highest price range:", averages.idxmax())
    print("Highest average:", round(averages.max(), 2))
```

```python
os.makedirs("../data/processed", exist_ok=True)

final_specification_summary.round(2).to_csv(
    "../data/processed/price_range_specification_summary.csv"
)

print("Specification summary saved successfully!")
```
